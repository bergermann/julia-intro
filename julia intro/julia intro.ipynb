## Concepts
### JIT compilation
Julia is Just-In-Time compiled, in others words at code execution time. This produces an overhead on first run-time (i.e. compilation time). This makes it slower compared to interpreted languages like Python for short programs, but faster on repeated execution/longer programs. Julias compiler is written in C, everything is native to Julia itself. <br/><br/>

### Type-System
Julia features a proper and important type-system:  
https://discourse.julialang.org/t/diagram-with-all-julia-types/5018<br/><br/>
Types can be arbitrarily combined in composite data types called `struct` (which itself are types again).<br/><br/>

### No Object-Orientation
Julia does not have classes, methods are not owned by classes, but by types. <br/><br/>

### Multiple-Dispatch
The managing process/paradigm to find the appropriate function for the input types is called *Multiple-Dispatch* which is the MVP for Julia.  
![md](multiple_dispatch.jpg)

### Dynamic and Static programming
Types can be left to be inferred by the compiler, or annotated to keep them static.

## Syntax
Very similar to Python for starters. Notable exceptions:
- no double dots `:` to declare a new block
- blocks are terminated with `end`
- functions are defined with `function` rather than `def`
- exponents are written as `x^a` instead of `x**a`.  
- `"string"` denotes strings, `'a'` only denotes single characters. `'string'` is invalid!
- Julia indexing starts with 1! First element is `array[1]`

Exempli gratia:
| Python | Julia |
|---|---|
|`def my_function(x,y,a=3):`<br/>----`a**=2`<br/>----`print(a)`<br/>----`print(x,",",y)`<br/>----`return x+y`|`function my_function(x,y; a=3)`<br/>----`a ^= 2`<br/>----`println(a)`<br/>----`println(x,",",y)`<br/>----`return x+y`<br/>`end`|  

Julia does NOT require indentation. Still helps readability. Two statements on one line can be seperated with `;`.  
Per convention keyword arguments are seperated with `;` from arguments.  
`println` prints a newline after the arguments, `print` does not.  
Equivalent: `println(x)` = `print(x,'\n')`<br/><br/>
Comments are written by `#`  
or multi-line  
`#=` <br/>
`some stuff`<br/>
`that would be commented out`<br/>
`=#`<br/><br/>
*Mutating* functions are denoted by a `!`, e.g. `push!(Array,x)`

### 1. Exercise: Quicksort implementation
https://en.wikipedia.org/wiki/Quicksort
```
algorithm quicksort(A, lo, hi) is 
  if lo >= hi or lo < 0 then 
    return
    
  p := partition(A, lo, hi) 
      
  quicksort(A, lo, p - 1) // Left side of pivot
  quicksort(A, p + 1, hi) // Right side of pivot

algorithm partition(A, lo, hi) is 
  pivot := A[hi] // Choose the last element as the pivot

  i := lo - 1

  for j := lo to hi - 1 do 
    if A[j] <= pivot then 
      i := i + 1
      swap A[i] with A[j]

  i := i + 1
  swap A[i] with A[hi]
  
  return i
  ```

### Composite types: `Structs`

````julia
# immutable, cannot be changed later

struct MyStructA
    field1::String  # data containers inside struct, can be named arbitrarily
    field2::Int     # fields SHOULD be type annotated for efficiency
    field3::Array{Float64}
end

# instantiate
a = MyStructA("Some string",5,zeros(5))
b = MyStructA("something else",8,zeros(5))
````

````julia
# access fields with fieldnames
a.field1
````

````julia
# or by symbol
x = :field2
getfield(b,x)
````

````julia
# fields of (immutable) structs cannot be changed
a.field2 = 3
````

````julia
# mutable containers of (immutable) structs CAN still be changed
a.field3[1] = 1
````

````julia
# mutable, can be changed later

abstract type MyStructs end # define abstract type as node in typegraph

mutable struct MyStructB <: MyStructs # declare hierarchy with <:
    field1::String
    field2::Float64
end

# instantiate, struct definition defines implict constructor function
b = MyStructB("Some other string",42.,zeros(5))
````

````julia
# check type of any variable with typeof function
typeof(b)
````

````julia
typeof(b) == MyStructB  # check if variable is specific type
b isa MyStructs         # better with isa keyword
````

````julia
# check subtype/supertype dependence with <: or isa
typeof(b) <: MyStructs
b isa MyStructs
````

````julia
# list suptypes/supertypes
supertypes(MyStructB)
subtypes(MyStructs)
````

````julia
# define costum constructors

mutable struct MyStructC <: MyStructs
    field1::String
    field2::Complex

    # inner constructor, overwrites default implicit constructor
    function MyStructC()
        new("default string",0+0im)
    end

    # this is equivalent to the default implicit constructor
    function MyStructC(str::String,c::Complex)
        new(str,c)
    end
end

# outer constructor, has to call inner constructor
function MyStructC(str::String,real::Real,imaginary::Real)
    MyStructC(str,real+1im*imaginary)
end

c = MyStructC()
````

````julia
methods(MyStructC)
````

*Variable shadowing* and *pass-by-value*  
Big pit-fall, be aware! When assignments are made inside a function every variable with same name from a higher scope and argument names will be *shadowed*, i.e. it will be ignored and the new assignment takes over the name space. Use mutating functions and statements to avoid.  

Further, container variables (e.g. arrays) are *passed-by-reference* i.e. a function that mutates its input array, mutates the real array.  
This is NOT the case for non-container variables (e.g. plain number). Reassign the variable with the return value instead.

````julia
a = zeros(5,5)

function nonMutatingF!(x::Array)
    # this does NOT change original x!
    # new local variable x is assigned, outer x is "shadowed"
    x = rand(5,5) # assignement statement
end

nonMutatingF!(a)
a
````

````julia
a = zeros(5,5)

function mutatingF!(x::Array)
    # operations operating explicitly on elements mutate outer x
    x .= rand(5,5) # assignment of elements (due to element-wise .=)
end

mutatingF!(a)
a
````

````julia
a = zeros(5,5)

function mutatingF!(x::Array)
    # slicing operates explicitly on elements
    x[:] = rand(5,5)
end

mutatingF!(a)
a
````

````julia
# integers variables can also be shadowed
b = 5

function testF!(x::Int)
    x = 8
end

testF!(b)
b
````

````julia
# however, non-container variables such as integers can NOT
# be mutated by passing them to functions
b = 5

function testF!(x::Int)
    x += 3
end

testF!(b)
b
````

````julia
# solution: mutate passed variables, reassing function return
b = 5

function testF(x::Int)
    return x+3
end

b = testF(b)
b
````

````julia
# number assignments copy the original
a = 5
b = a
a = 3
# b does not get changed after a was changed, b = a creates indepentend copy
println(b)

b = 10
# or vice versa
println(a)
````

````julia
# array assignments produce a reference
a = [1 2
    3 5]

b = a
# a gets changed after b gets changed, b and a refer to same array
# use copy(a) to make explicit indepentend copy
b[1] = 0

a,b
````

### Performance Tips
- predefine arrays
- use mutation where applicable
- avoid "type instability"
- use type annotations inside struct definitions, arrays -> if possible, parametric types
- avoid global variables, run code in functions
- respect row priority over colums when iterating through matrices
- use `for i in eachindex(A)` instead of `for i in 1:length(A)` if applicable
- read https://docs.julialang.org/en/v1/manual/performance-tips/

````julia
@time begin
    a = []

    for i in 1:10000
        push!(a,i)
    end
end

@time begin
    a = Array{Int}(undef,1000)

    for i in 1:1000
        a[i] = i
    end
end
````
